```python
import pandas as pd 
import numpy as np 
```

```python
data=r"C:\Users\Mahakaal\Documents\churnsense\data\raw\Telco-Customer-Churn dataset.csv"
df=pd.read_csv(data)
```

```python
df.info()
```

```python
df.columns = df.columns.str.lower().str.replace(' ', '_')
df.columns
```

```python
string_columns = df.select_dtypes(include=['object']).columns.tolist()

for col in string_columns:
    df[col] = df[col].astype(str).str.lower().str.strip().str.replace(' ', '_')

df[string_columns].head()
```

```python
df['totalcharges'] = pd.to_numeric(df['totalcharges'], errors='coerce')
df['totalcharges'] = df['totalcharges'].fillna(df['totalcharges'].median())
```

```python
binary_cols = ['partner', 'dependents', 'phoneservice', 'paperlessbilling', 'churn']
binary_map = {'yes': 1, 'no': 0}

for col in binary_cols:
    if col in df.columns:
        df[col] = df[col].map(binary_map).fillna(df[col]).astype(int)

df.info()
```

```python
df.to_csv(r"C:\Users\Mahakaal\Documents\churnsense\data\processed\Telco-Customer-Churn dataset_cleaned.csv", index=False)
```
